# SECTION 1: PRODUCTS END POINT

```python
import requests
```

```python
products_url = "https://dummyjson.com/products"
```

```python
products_data = requests.get(products_url)
```

```python
#check response
```

```python
products_data.status_code
```

```python
#Extract the JSON response
```

```python
products_json = products_data.json()
```

```python
#check data type
```

```python
#check how many products
```

```python
#Convert to DataFrame
```

```python
import pandas
```

```python
df = pandas.DataFrame(products_json['products'])
```

```python
#check the first 3 columns
```

```python
df.head(3)
```

```python
#Save as CSV
```

```python
df.to_csv("products_data.csv", index=False)
```

```python
#verify the csv was saved
```

```python
import pandas
check = pandas.read_csv("products_data.csv")
print(f"✅ File saved successfully!")
print(f"Shape: {check.shape}")
check.head()
```
